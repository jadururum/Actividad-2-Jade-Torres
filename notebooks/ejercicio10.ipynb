```python
nombres=['Valentina','Mateo','Camila','Santiago','Lucía']
stats = {n: {'total': 0, 'ganadas': 0, 'mejor': 0} for n in nombres}
```

```python
for i,r in enumerate(rounds,1):
    print(f"\nRonda {i} - {r['theme']}:")

    puntos_ronda = ejercicio10.puntos_por_ronda(r['scores'])
    ganador, puntos_ganador = ejercicio10.obtener_ganador(puntos_ronda)
    
    print(f"  Ganador: {ganador} ({puntos_ganador} pts)")
    
    print(f"  {'Cocinero':15} | {'Puntos':10}") 
    print("  " + "-" * 28)
    
    tabla_actual=sorted(puntos_ronda.items(),key=lambda x: x[1],reverse= True )
    for nombre, puntos in tabla_actual:
        print(f" {nombre:20}{puntos}")
    
    for nombre, puntos in puntos_ronda.items():
        stats[nombre]['total'] += puntos
        if puntos > stats[nombre]['mejor']:
            stats[nombre]['mejor'] = puntos

    stats[ganador]['ganadas'] += 1
            
    
```
