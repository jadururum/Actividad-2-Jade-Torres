```python
nombres=['Valentina','Mateo','Camila','Santiago','Lucía']
stats = {n: {'total': 0, 'ganadas': 0, 'mejor': 0} for n in nombres}
```

```python
for i,r in enumerate(rounds,1):
    print(f"\nRonda {i} - {r['theme']}:")

    puntos_ronda = ejercicio10.puntos_por_ronda(r['puntajes'])
    ganador, pts_ganador = ejercicio10.obtener_ganador(puntos_ronda)
    
    print(f"  Ganador: {ganador} ({pts_ganador} pts)")
    
    stats[ganador]['ganadas'] += 1
    
    for nombre, puntos in puntos_ronda.items():
        stats[nombre]['total'] += puntos
        if puntos > stats[nombre]['mejor']:
            stats[nombre]['mejor'] = puntos
            
    print("  ... (tabla de posiciones de la ronda)")
```

```python
print("\n" + "="*30)
print("TABLA DE POSICIONES FINAL")
print(f"{'Cocinero':12} {'Puntaje':8} {'Ganadas':10} {'Mejor':8} {'Promedio':8}")
print("-" * 55)

ranking = sorted(stats.items(), key=lambda x: x[1]['total'], reverse=True)

for nombre, data in ranking:
    promedio = data['total'] / 5
    print(f"{nombre:12} {data['total']:<8} {data['ganadas']:<10} {data['mejor']:<8} {promedio:.1f}")
```
